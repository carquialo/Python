### Las tuplas

Las tuplas van entre paréntesis (), y opcional sin nada pero desaconsejable. 

```python
mitupla = ("Pepe", 18, 9, 2000)
print(mitupla)
```

```python
print (mitupla[1])
```

```python
#Podemos convertir una tupla en una lista con list()

mitupla = ("Pepe", 18, 9, 2000)
milista = list(mitupla)

print(milista) # Cuando lo imprimimos si nos fijamos vemos el resultado como lista (está entre corchetes)
```

```python
# Podemos hacerlo también al revés, es decir, convertir la lista en tupla con tuple()

milista = ["Pepe", 18, 9, 2000]
mitupla= tuple(milista)

print(mitupla)
```
