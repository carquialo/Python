### Condicionales 3

Concatenación de condicionales de comparación. 


```python
edad = 9

if edad < 100:
    print("Edad correcta")
else:
    print("Edad incorrecta")
```

```python
# Vamos a comparalo con otro menor que.
edad = -9       # En este caso pondremos un número negativo. 
if 0 < edad < 100:
    print("Edad correcta")
else:
    print("Edad incorrecta")
```
