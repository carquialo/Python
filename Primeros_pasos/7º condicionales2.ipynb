### Condicionales 2

En este apartado veremos la sentencias else y elif.

```python
# Con "else" podemos traducirlo como "y si no es verdad".

print("Verificación de usuario")

edad_usuario = int(input("Introduzca tu edad, por favor"))

if edad_usuario >= 18:
    print("Tienes acceso, puedes entrar.")
else: 
    print("No tiene acceso.")
```

```python
print("Verificación de usuario")

nota = int(input("Introduzca tu nota, por favor"))

if nota < 5: 
    print("Insuficiente")
elif nota < 6: 
    print("Suficiente")
elif nota<7: 
    print("Bien")
elif nota<9 :
    print("Notable")
else:
    print("Sobresaliente")
```
