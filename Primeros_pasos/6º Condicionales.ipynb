### Condicionales

```python
# Usaremos la condicional if.
def evaluacion (nota):
    valoracion = "Aprobado"
    if nota<5 :
        valoracion="Suspenso"
    return valoracion
print(evaluacion(4))
```

```python
# Crearemos un mini programa en el que introduciremos una nota y nos devolverá la valoración

print("La valoración del alumno es : ")
nota_alumno=int(input())

def evaluacion (nota):
    valoracion = "Aprobado"
    if nota<5 :
        valoracion="Suspenso"
    return valoracion
print(evaluacion(nota_alumno))
```
