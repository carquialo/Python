### Diccionario

```python
# La sintaxis es la siguiente: diccionario = {clave : valor}.

midiccionario = {"España":"Madrid", "Francia":"París", "Alemania":"Berlín"}

print(midiccionario)
```

```python
# Si quisiera imprimir el valor de una clave haríamos lo siguiente: 

print(midiccionario["España"])  # Habría que poner entre corchete la clave.
```

```python
# Podemos hacer un diccionario dentro de otro. 

midiccionario = {23:"Pepe", "Nombre": "Pepito", "Equipo":"San Agustín","anillos":{"temporadas":[1992,1993,1994]}}
print(midiccionario["anillos"])
```

```python
# Tenemos unas formas para que nos devuelva las claves que es de la siguiente forma:

print(midiccionario.keys())
```

```python
# Otra forma es que nos imprima los valores : 

print(midiccionario.values())
```

```python
# La longitud de nuestro diccionario.
print(len(midiccionario))
```
