### For

¿Para qué se usa el bucle for en python? 

Se usa para iterar una secuencia de elementos, cadenas de texto, listas, tuplas, diccionarios, ect. Es útil para cuando quieres repetir varias veces un código. 

```python
# Esto sería una lista. 

frutas = ["manzana", "plátano", "pera" ]

for fruta in frutas: 
    print(fruta)
```

```python
# En este ejemplo usaremos range() para iterar un número específico de veces.

for i in range(6):
    print(i)

```

```python
# Otro ejemplo sería de un diccionario. 

edades = {"Carlos":45, "Pepa":24, "Roberta":23}

for nombre, edad in edades.items():
    print(f"{nombre} tiene {edad} años")
```
